# Lab II — Same matrix, three passports

The **same array of entries** lives in different scalar worlds. Be precise about restriction versus extension of scalars.

```python
from lasensei.finite import F2,F4,F8
from lasensei.linear import power,eye,vec
from lasensei.lab_registry import passports
passports()
```

### Passport 1: a binary operator

The three nonzero vectors cycle. There is no nonzero fixed eigenvector, and $A^3=I$.

```python
A=[[0,1],[1,1]]
v=[1,0]
orbit=[]
for _ in range(3):
    orbit.append(v)
    v=vec(A,v,F2)
assert v==[1,0] and power(A,3,F2)==eye(2)
orbit
```

### Passport 2: multiplication by $\alpha$ on $\mathbb F_4$

```python
alpha=2  # encoded as 0+1*alpha
[F4.display(F4.mul(alpha,x)) for x in (1,alpha,F4.mul(alpha,alpha))]
```

### Passport 3: scalar extension produces two eigenvectors

```python
roots=[x for x in range(4) if F4.add(F4.add(F4.mul(x,x),x),1)==0]
assert roots==[2,3]
for x in roots:
    e=[1,x]
    assert vec(A,e,F4)==[F4.mul(x,z) for z in e]
[(F4.display(x), [F4.display(z) for z in [1,x]]) for x in roots]
```
